```python
import cv2
import numpy as np

main_img = cv2.imread('m1.png',cv2.IMREAD_GRAYSCALE)
template = cv2.imread('template.png',cv2.IMREAD_GRAYSCALE)

if main_img is None or template is None:
    print("Could not open or find the images!")
else:
    results = cv2.matchTemplate(main_img, template, cv2.TM_CCOEFF_NORMED)
    min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(results)
   
    confidence = 0.8
    if max_val >= confidence:
        print("พบไอเทม")
        
        main_image_color = cv2.imread('m1.png')
        h, w = template.shape
       
        #หาค่าด้านล่างขวา
        bottom_right = (max_loc[0] + w, max_loc[1] + h)
        
        cv2.rectangle(main_image_color, max_loc, bottom_right, (0, 255, 248), 3)
        
        cv2.imshow('Detected', main_image_color)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("ไม่พบไอเทม")
        
        
         
        
    
    
```

```python
import win32api,win32con,win32gui
import cv2
import numpy as np

get_hwnd =  win32gui.FindWindow(None, "MapleStory")
left,top,rigth,bottom = win32gui.GetWindowRect(get_hwnd)

width = rigth - left
height = bottom - top
# สร้างDC
hwnd_dc = win32gui.GetWindowDC(get_hwnd)
mfc_dc = win32gui.CreateDCFromHandle(hwnd_dc)
save_dc = mfc_dc.CreateCompatibleDC()
# สร้างbitmap 
bitmap = win32gui.CreateBitmap()
bitmap.CreateCompatibleBitmap(mfc_dc, width, height)
save_dc.SelectObject(bitmap)


bitmap_info = bitmap.GetInfo()
bitmap_str = bitmap.GetBitmapBits(True)
img = np.frombuffer(bitmap_str, dtype='uint8').reshape((bitmap_info['bmHeight'], bitmap_info['bmWidth'], 4))
img = img[...,:3]
img = np.ascontiguousarray(img)

win32gui.DeleteObject(bitmap.GetHandle())
save_dc.DeleteDC()
mfc_dc.DeleteDC()
win32gui.ReleaseDC(get_hwnd, hwnd_dc)





```
